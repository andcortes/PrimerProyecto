- Crea un notebook `panel.ipnyb` donde implementes estas funcionalidades:
	- el producto con más ingresos
	- filtra las ventas por un vendedor concreto
	- calcula el ticket medio

```python
import json
from pathlib import Path

ruta_datos = Path("data.json")
if not ruta_datos.exists():
    ruta_datos = Path("Proyecto_TTAD_1") / "data.json"

with ruta_datos.open(encoding="utf-8") as archivo:
    datos = json.load(archivo)

ventas = datos["ventas"]
print(f"Se han cargado {len(ventas)} ventas")
```

Funcionalidad 1.

```python
ingresos_por_producto = {}

for venta in ventas:
    producto = venta["producto"]
    ingreso = venta["cantidad"] * venta["precio_unitario"]
    ingresos_por_producto[producto] = ingresos_por_producto.get(producto, 0) + ingreso

producto_mas_ingresos = max(
    ingresos_por_producto,
    key=ingresos_por_producto.get
)

print("Producto con más ingresos:", producto_mas_ingresos)
print("Ingresos:", ingresos_por_producto[producto_mas_ingresos])
```

Funcionalidad 2.

```python
vendedor = "Marta"

ventas_vendedor = [
    venta for venta in ventas
    if venta["vendedor"] == vendedor
]

print(f"Ventas de {vendedor}: {len(ventas_vendedor)}")
for venta in ventas_vendedor:
    print(venta)
```

Funcionalidad 3.

```python
ingreso_total = sum(
    venta["cantidad"] * venta["precio_unitario"]
    for venta in ventas
)

ticket_medio = ingreso_total / len(ventas)

print("Ingreso total:", ingreso_total)
print("Ticket medio:", round(ticket_medio, 2))
```
